```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_companies_of_Nigeria'

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36'
}
page = requests.get(url, headers=headers)
if page.status_code == 200:
    soup = BeautifulSoup(page.text, 'html.parser')
    print("Success! Soup is ready.")
    print(soup.title.text)
else:
    print(f"Error:Received status code {page.status_code}")
```

```python
soup.find('table', class_ = "wikitable sortable")
```

```python
table = soup.find('table', class_ = 'wikitable')
```

```python
print(table)
```

```python
industries_titles = table.find_all('th')
```

```python
industries_titles
```

```python
industries_tables_titles = [title.text.strip() for title in industries_titles if title.text.strip()]
```

```python
print(industries_tables_titles)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns = industries_tables_titles)

df
```

```python
column_data = table.find_all('tr')
```

```python
for row in column_data:
    row_data = row.find_all('td')
    individual_row_data = [data.text.strip() for data in row_data]
    
    if len(individual_row_data) >= len(df.columns):
        length = len(df)
        df.loc[length] = individual_row_data[:len(df.columns)]
```

```python
df
```

```python
df.to_csv(r'C:\Users\DELL\Documents\Web scraping\Industries.csv', index = False)
```

```python
# --- CLEANING SECTION ---

df.columns = ['Name', 'Industry', 'Sector', 'Headquarters', 'Founded', 'Notes', 'Status']

df['Founded'] = pd.to_numeric(df['Founded'], errors='coerce')

df['Status'] = df['Status'].str.strip().str.upper()

df.dropna(subset=['Name'], inplace=True)

print("Data cleaning complete. Ready for visualization.")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

main_color = "steelblue"
sns.set_theme(style="whitegrid")
plt.figure(figsize=(15, 12))

plt.subplot(2, 2, 1)
sns.countplot(y='Industry', data=df, 
              order=df['Industry'].value_counts().index, 
              color=main_color)
plt.title('Distribution of Companies by Industry', fontsize=13)
plt.xlabel('Number of Companies')

plt.subplot(2, 2, 2)
status_counts = df['Status'].value_counts()
if not status_counts.empty:
    labels = ['Private (P)' if x == 'P' else 'State (S)' for x in status_counts.index]
    plt.pie(status_counts, labels=labels, autopct='%1.1f%%', 
            colors=['#4682B4', '#A9A9A9'], startangle=140)
    plt.title('Ownership: Private vs State-Owned')

plt.subplot(2, 2, 3)
sns.histplot(df['Founded'].dropna(), bins=15, kde=True, color=main_color)
plt.title('Company Foundation Timeline', fontsize=13)
plt.xlabel('Year')

plt.subplot(2, 2, 4)
top_cities = df['Headquarters'].value_counts().head(5)
sns.barplot(x=top_cities.values, y=top_cities.index, color=main_color)
plt.title('Top 5 Business Hubs in Nigeria', fontsize=13)
plt.xlabel('Number of Headquarters')

plt.tight_layout()
plt.show()
```
